```python
import numpy as np
import pandas as pd

#Series-block
```

```python
#Series with label-based indexing, without index< numering starts from 0
named_ages = pd.Series(data=[22, 35, 25, 16], index=["Aline", "Beatrice", "Clara", "Diana"], dtype=np.int32, name="age")
print(named_ages)
print(named_ages["Beatrice"])
print(named_ages.loc["Beatrice"])
print(named_ages.iloc[1])
```

```python
s = pd.Series(data=[0, 10, 20, 30], index=[1, 2, 3, 4])
print(s)
s.pop(1) # удаление элемента с индексом-меткой 1
print(s)
```

```python
#indexes are not unique
a = pd.Series(data=[10, 20, 40], index=[1, 2, 4])
b = pd.Series(data=[100, 200, 300, 200, 100], index=[1, 2, 3, 4, 5])
c = pd.concat([a, b])
print(c.index)
print(c.index.is_unique)
print(c.unique())
#frequency:
print("freq:", c.value_counts())
#count NaN type
print("Nan:", c.isna().sum())
#New series without Nan, c is still the same
print("Non-Nan", c.dropna())
```

```python
A = pd.Series(data=[100, 200, 500, 0], index=["Aline", "Beatrice", "Clara", "Diana"], name="money")
B = pd.Series(data=[300, 0], index=["Diana", "Aline"])
print(A)
A.update(B)
print(A)
```

```python
#Data-frame-block
df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6], "C": [7, 8, 9]},
                  index = ["first", "second", "third"])
print(df)
print("\nAsk for Series\n")
print(df['A'])
print(df.A)
#but with point i cannot do smth like df.D = df.A*100 + df.B*10 + df.C
#A comma inside square brackets allows you to retrieve the value of a specific cell, or a slice across both rows and columns simultaneously:
print("\ncross-section\n", df.loc['first':'second', 'B':'C'])

print("\nconvolution\n", df.aggregate(['min', 'idxmin', 'max', 'idxmax', 'mean', 'sum']))
```

```python
commits = pd.DataFrame({"name": ["Петя", "Вася", "Таня", "Петя", "Петя", "Таня", "Вася", "Петя", "Таня"],
                        "potatoes": [3, 4, 0, 3, 4, 0, 6, 3, 0],
                        "carrots": [0, 2, 5, 2, 1, 4, 15, 2, 6]})
print(commits)
print("\nGrouped by\n", commits.groupby("name").sum())
print("\nAgr:\n", commits.groupby("name").aggregate(["min", "max"]))
```
